```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('Model is ready to use!')
```

```python
# STEP-1: Define the state
from typing import TypedDict, Annotated
from langchain_core.messages import AnyMessage
from langgraph.graph.message import add_messages

class StoryState(TypedDict):
    stories: Annotated[list[AnyMessage], add_messages]

print('Story state is ready..')
```

```python
# STEP-2: Define the nodes
def narrate(state: StoryState) -> dict:
    """Just telling Marvel stories"""
    response = model.invoke(state['stories'])
    return {"stories": [response]}

print('Graph Nodes are ready')
```
