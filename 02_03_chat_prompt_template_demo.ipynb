```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('Model is ready to use!')
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country}?",
    input_variables=["country"]
)
prompt = prompt_template.format(country="France")

response = model.invoke(prompt)
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate([
    ("user", "What is the capital of {country}?")
])

prompt = chat_prompt_template.invoke({"country": "France"})
response = model.invoke(prompt)

print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate([
    ("user", "Explain the {theorem} in 3 lines")
])

prompt = chat_prompt_template.invoke({"theorem": "pythagoras theorem"})
response = model.invoke(prompt)

print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate([
    ("user", "Explain the {theorem} in 3 lines"),
    ('system', "You are an expert mathematician and will provide a concise explanation of the theorem.")
])

prompt = chat_prompt_template.invoke({"theorem": "pythagoras theorem"})
response = model.invoke(prompt)

print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate([
    ("user", "Explain the {theorem} "),
    ('system', "You are an illiterate person and no idea about mathematics")
])

prompt = chat_prompt_template.invoke({"theorem": "pythagoras theorem"})
response = model.invoke(prompt)

print(response)
```
