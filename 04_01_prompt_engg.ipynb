```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('Model is ready to use!')
```

```python
# Zero shot prmpting

from langchain_core.prompts import PromptTemplate

prompt_template =  PromptTemplate(
    template="Which is the capital of {country}?",
    input_variables=["country"]
)
prompt = prompt_template.format(country="France")

response = model.invoke(prompt)
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

one_shot_prompt = (
    f"Example:\n"
    f"Question: Translate 'Hello, how are you?' into Hindi.\n"
    f"Answer: Aap kaise ho?\n\n"
    f"Now answer this:\n"
    f"Question: Translate 'What is your name?' into Hindi.\n"
    f"Answer: \n"
)
prompt_template =  PromptTemplate(
    template=one_shot_prompt
)
prompt = prompt_template.format(language="Hindi")

response = model.invoke(prompt)
print(response)
```

```python
from langchain_core.output_parsers import StrOutputParser

string_output_parser = StrOutputParser()

prompt =  PromptTemplate(
    template="Write a poem on the city of Paris"
)

chain = prompt | model | string_output_parser

response = chain.invoke({})
print(type(response))
print(response)
```

```python
from langchain_core.output_parsers import ListOutputParser

class CustomListOutputParser(ListOutputParser):
    def parse(self, text: str) -> list:
        # Custom parsing logic to extract list items from the text
        items = [item.strip() for item in text.split('\n') if item.strip()]
        return items

prompt =  PromptTemplate(
    template="Please give me list of only names of top 5 countries by their economy size"
)

list_output_parser = CustomListOutputParser()
chain = prompt | model | list_output_parser

response = chain.invoke({})
print(type(response))
print(response)
```

```python
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel, Field

prompt = PromptTemplate(
    template="Extract the person details from the following text:\nJohn Doe, 30 years old, lives in New York.",
)

# output schema for te extracted details
class Person(BaseModel):
    name: str = Field("Name")
    age: int = Field("Age")
    city: str = Field("City")

json_output_parser = JsonOutputParser(pydantic_object=Person)

chain = prompt | model | json_output_parser

response = chain.invoke({})
print(type(response))
print(response)
```

```python
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel, Field

prompt = PromptTemplate(
    template="Extract the person details from the following text and " \
    "convert it into the format instructions {format_instructions}:\nJohn Doe, {age} years old, lives in New York.",
    input_variables=["age"],
    partial_variables=['format_instructions'], # this was last line written, to continue from here
)

# output schema for te extracted details
class Person(BaseModel):
    name: str = Field("Name")
    age: int = Field("Age")
    city: str = Field("City")

pydantic_output_parser = PydanticOutputParser(pydantic_object=Person)

chain = prompt | model | pydantic_output_parser

response = chain.invoke({})
print(type(response))
print(response)
```
