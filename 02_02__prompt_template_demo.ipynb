```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('Model is ready to use!')
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="What is the capital of {country}?",
    input_variables=["country"]
)
prompt = prompt_template.format(country="France")

response = model.invoke(prompt)
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="Give me brief information about the city {city} located in the country {country}",
    input_variables=["city", "country"]
)
prompt = prompt_template.format(city="Paris", country="France")

response = model.invoke(prompt)
print(response)
```
